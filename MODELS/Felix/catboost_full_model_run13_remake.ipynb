## 2. Add Built-In Toolkit Features

However to prevent any subtle leakage from the `dropna()` step interacting with

## 3. Add New Custom Features In The Notebook

This is where developers keep their freedom. The toolkit supplies the base feature set;
everything below is notebook-local and experimental.

We add three categories of custom features:

**Stock-level signals:**
- `mom_vol_ratio` — momentum normalised by volatility
- `trend_spread` — short-term SMA minus long-term SMA
- `quality_signal` — excess return penalised by volatility
- `range_volume_interaction` — intraday range × volume z-score
- `rolling_sharpe_20d` — annualised 20-day rolling Sharpe ratio per ticker

**Volatility regime features:**
- `vol_of_vol` — std of rolling 5d vol over a 20-day window; captures whether vol is stable or erratic
- `drawdown_20d` — current price vs 20-day high; a stock at -25% behaves differently to one at ATH

**Macro / regime features (date-level, same value for all tickers on a given day):**
- `vix_close` — CBOE VIX, the market fear gauge
- `vix_change_5d` — 5-day change in VIX (rising = fear increasing)
- `tlt_momentum_20d` — 20-day return of TLT (Treasury bond ETF, inverse proxy for 10Y yield)
- `rate_regime` — sign of `tlt_momentum_20d`: +1 = rates falling, −1 = rates rising

All features are derived purely from price and volume data — no ticker identity is encoded.
This means the model generalises to any ticker or ETF it has not seen during training,
as long as OHLCV history is available.


```python
frame = base_features.copy()
eps = 1e-6

# ── Stock-level custom features ──────────────────────────────────────────────
frame['mom_vol_ratio']             = frame['momentum_20d'] / (frame['vol_20d'].abs() + eps)
frame['trend_spread']              = frame['price_to_sma_20d'] - frame['price_to_sma_50d']
frame['quality_signal']            = frame['excess_return_20d_vs_spy'] - 0.5 * frame['vol_20d']
frame['range_volume_interaction']  = frame['intraday_range'] * frame['volume_zscore_20d']

# ── Volatility regime features ────────────────────────────────────────────────
daily_returns = (
    prices.sort_values(['ticker', 'date'])
    .assign(daily_ret=lambda df: df.groupby('ticker')['close'].pct_change())
)

# rolling_sharpe_20d
rolling_sharpe = (
    daily_returns.groupby('ticker')['daily_ret']
    .transform(lambda s: s.rolling(20).mean() / (s.rolling(20).std(ddof=0) + eps))
)
daily_returns['rolling_sharpe_20d'] = rolling_sharpe * (252 ** 0.5)
sharpe_lookup = daily_returns[['date', 'ticker', 'rolling_sharpe_20d']]
frame = frame.merge(sharpe_lookup, on=['date', 'ticker'], how='left')

# vol_of_vol
vol_of_vol = (
    daily_returns.groupby('ticker')['daily_ret']
    .transform(lambda s: s.rolling(5).std().rolling(20).std())
)
daily_returns['vol_of_vol'] = vol_of_vol
vov_lookup = daily_returns[['date', 'ticker', 'vol_of_vol']]
frame = frame.merge(vov_lookup, on=['date', 'ticker'], how='left')

# drawdown_20d
drawdown_20d = (
    prices.sort_values(['ticker', 'date'])
    .groupby('ticker')['close']
    .transform(lambda s: s / s.rolling(20).max() - 1)
)
drawdown_lookup = (
    prices.sort_values(['ticker', 'date'])[['date', 'ticker']]
    .assign(drawdown_20d=drawdown_20d.values)
)
frame = frame.merge(drawdown_lookup, on=['date', 'ticker'], how='left')

# ── Macro / regime features ───────────────────────────────────────────────────
import yfinance as yf
_start = '2014-01-01'
_end   = '2025-12-31'

def _fetch_close(ticker, start, end):
    df = yf.download(ticker, start=start, end=end, progress=False, auto_adjust=True)
    close = df['Close']
    if isinstance(close, pd.DataFrame):
        close = close.iloc[:, 0]
    return close

vix_raw = _fetch_close("^VIX", _start, _end).rename("vix_close")
tlt_raw = _fetch_close("TLT",  _start, _end).rename("tlt_close")

macro = (
    pd.concat([vix_raw, tlt_raw], axis=1)
    .reset_index()
    .rename(columns={'Date': 'date', 'Datetime': 'date', 'index': 'date'})
    .assign(date=lambda df: pd.to_datetime(df['date']).dt.normalize())
)
macro['vix_change_5d']    = macro['vix_close'].pct_change(5)
macro['tlt_momentum_20d'] = macro['tlt_close'].pct_change(20)
macro['rate_regime']      = macro['tlt_momentum_20d'].apply(lambda x: 1.0 if x >= 0 else -1.0)
macro_feature_names = ['vix_close', 'vix_change_5d', 'tlt_momentum_20d', 'rate_regime']
macro = macro[['date'] + macro_feature_names].dropna()
frame['date'] = pd.to_datetime(frame['date']).dt.normalize()
frame = frame.merge(macro, on='date', how='left')
print('Macro features merged.')

# ── Regime & categorical features ────────────────────────────────────────────
frame['vix_regime'] = pd.cut(
    frame['vix_close'],
    bins=[0, 15, 20, 30, 100],
    labels=[0, 1, 2, 3]
).astype(float)

frame['momentum_regime'] = pd.cut(
    frame['momentum_20d'],
    bins=[-np.inf, -0.05, 0, 0.05, np.inf],
    labels=[0, 1, 2, 3]
).astype(float)

eps2 = 1e-6
daily_ret_std5  = daily_returns.groupby('ticker')['daily_ret'].transform(lambda s: s.rolling(5).std())
daily_ret_std20 = daily_returns.groupby('ticker')['daily_ret'].transform(lambda s: s.rolling(20).std())
vol_expansion = daily_ret_std5 / (daily_ret_std20 + eps2)
vol_exp_lookup = (
    daily_returns[['date', 'ticker']]
    .assign(vol_expansion_ratio=vol_expansion.values)
)
frame = frame.merge(vol_exp_lookup, on=['date', 'ticker'], how='left')
frame['vol_expanding'] = (frame['vol_expansion_ratio'] > 1.2).astype(float)

print('Regime and categorical features added.')

# ── Cross-sectional rank features (computed AFTER split to avoid leakage) ───
# These rank each stock relative to all others on the SAME date.
# IMPORTANT: ranks are computed separately on train, val, test to avoid
# future dates influencing the rank distribution of past dates.
# We add them to the frame now as NaN placeholders — actual values filled in cell 15.
frame['xs_return_rank']   = np.nan  # rank of return_20d across tickers per date
frame['xs_momentum_rank'] = np.nan  # rank of momentum_20d across tickers per date
frame['xs_sharpe_rank']   = np.nan  # rank of rolling_sharpe_20d across tickers per date
frame['xs_quality_rank']  = np.nan  # rank of quality_signal across tickers per date
print("Cross-sectional rank placeholders added — will be filled per split in cell 15.")

custom_feature_names = [
    'mom_vol_ratio',
    'trend_spread',
    'quality_signal',
    'range_volume_interaction',
    'rolling_sharpe_20d',
    'vol_of_vol',
    'drawdown_20d',
    'vix_close',
    'vix_change_5d',
    'tlt_momentum_20d',
    'rate_regime',
    'xs_return_rank',
    'xs_momentum_rank',
    'xs_sharpe_rank',
    'xs_quality_rank',
    'vix_regime',
    'momentum_regime',
    'vol_expansion_ratio',
    'vol_expanding',
]

all_feature_names = base_feature_names + custom_feature_names
print(f'Total features: {len(all_feature_names)} ({len(base_feature_names)} base + {len(custom_feature_names)} custom)')
display(frame.loc[:, ['date', 'ticker'] + custom_feature_names].head())
```

```python
# ── Build targets for multiple horizons ──────────────────────────────────────
return_targets_5d  = make_forward_return_target(prices, horizon=5)
return_targets_10d = make_forward_return_target(prices, horizon=10)
return_targets_21d = make_forward_return_target(prices, horizon=21)

alpha_targets_5d   = make_forward_alpha_target(prices, horizon=5)
alpha_targets_10d  = make_forward_alpha_target(prices, horizon=10)
alpha_targets_21d  = make_forward_alpha_target(prices, horizon=21)

vol_targets_5d     = make_forward_realized_vol_target(prices, window=5)
vol_targets_10d    = make_forward_realized_vol_target(prices, window=10)
vol_targets_21d    = make_forward_realized_vol_target(prices, window=21)

# Merge all target horizons into the feature frame
target_frame = frame.copy()
for df in [
    return_targets_5d, return_targets_10d, return_targets_21d,
    alpha_targets_5d,  alpha_targets_10d,  alpha_targets_21d,
    vol_targets_5d,    vol_targets_10d,    vol_targets_21d,
]:
    target_frame = target_frame.merge(df, on=['date', 'ticker'], how='left')

# ── Drop NaNs excluding rank columns ─────────────────────────────────────────
# xs_*_rank columns are NaN placeholders at this stage — they get filled
# per split in cell 15 to prevent leakage. Exclude them from dropna().
rank_cols = [c for c in target_frame.columns if c.startswith("xs_")]
cols_to_check = [c for c in target_frame.columns if c not in rank_cols]

target_frame = (
    target_frame
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=cols_to_check)
    .reset_index(drop=True)
)

# ── Active target columns ─────────────────────────────────────────────────────
return_target_col = f"forward_return_{horizon}d"
alpha_target_col  = f"forward_alpha_{horizon}d_vs_spy"
vol_target_col    = f"forward_realized_vol_{horizon}d"

print(f"Active horizon: {horizon}d")
print(f"Return target:  {return_target_col}")
print(f"Alpha target:   {alpha_target_col}")
print(f"Vol target:     {vol_target_col}")
print("Modeling frame shape after dropping nulls:", target_frame.shape)
print("Rank cols still NaN (expected):", target_frame[rank_cols].isna().all().all())
display(target_frame.head())
```

## 5. Shared Train / Validation / Test Splits And Feature Scaling

This section does two very important things:

1. **Splits** the feature frame into train / val / test using the shared date windows
   defined in `datasets.toml` — so every model in the repo is evaluated on the same periods.

2. **Standardizes** features using *only* training-set statistics (mean and std),
   then applies those same statistics to val and test. This prevents data leakage.

3. **Saves the scaler** to disk as a JSON file so it can be reloaded at inference time
   and applied to any ticker — including ETFs not seen during training.
   Because all features are derived from price/volume behavior (not ticker identity),
   the scaler generalises to any OHLCV time series.


```python
import json as _json

# ── Cross-sectional rank function (defined first so it can be called below) ──
def add_xs_ranks(df):
    """Add cross-sectional percentile ranks per date within this split only.
    Computed per split to prevent leakage across train/val/test boundaries.
    """
    df = df.copy()
    for src_col, rank_col in [
        ('return_20d',         'xs_return_rank'),
        ('momentum_20d',       'xs_momentum_rank'),
        ('rolling_sharpe_20d', 'xs_sharpe_rank'),
        ('quality_signal',     'xs_quality_rank'),
    ]:
        if src_col in df.columns:
            df[rank_col] = df.groupby('date')[src_col].rank(pct=True)
    return df

# ── Override train cutoff to 2020-12-31 ──────────────────────────────────────
# Ensure date column is datetime before filtering
target_frame["date"] = pd.to_datetime(target_frame["date"])

train = target_frame[
    (target_frame["date"] >= pd.Timestamp("2014-01-02")) &
    (target_frame["date"] <= pd.Timestamp("2020-12-31"))
].copy()
# Embargo: skip the first EMBARGO_DAYS trading days after train cutoff
embargo_end = pd.Timestamp("2020-12-31") + pd.Timedelta(days=EMBARGO_DAYS * 2)
val = target_frame[
    (target_frame["date"] >= embargo_end) &
    (target_frame["date"] <= pd.Timestamp("2021-12-31"))
].copy()
test = target_frame[
    (target_frame["date"] >= pd.Timestamp("2022-01-03")) &
    (target_frame["date"] <= pd.Timestamp("2025-12-31"))
].copy()

# Apply ranks per split (no leakage)
train = add_xs_ranks(train)
val   = add_xs_ranks(val)
test  = add_xs_ranks(test)
print("Cross-sectional ranks computed per split.")
print("Overridden splits:")
print("Train rows:", len(train), "| cutoff: 2020-12-31")
print("Val rows:  ", len(val))
print("Test rows: ", len(test))

# ── Rebalancing frequency subsampling ────────────────────────────────────────
def get_rebalance_dates(df, every_n_days):
    """Return every Nth unique date from a frame."""
    dates = sorted(df["date"].unique())
    return dates[::every_n_days]

rebalance_dates = get_rebalance_dates(test, rebalance_every_n_days)
test_rebalance  = test[test["date"].isin(rebalance_dates)].copy()
print(f"\nRebalance every {rebalance_every_n_days} days")
print(f"Full test rows: {len(test)} → Rebalance test rows: {len(test_rebalance)}")
print(f"Number of rebalance dates: {len(rebalance_dates)}")

# ── Build scaler from training data only (no leakage) ────────────────────────
train_means = train[all_feature_names].mean()
train_stds  = train[all_feature_names].std(ddof=0).replace(0.0, 1.0)

output_dir.mkdir(parents=True, exist_ok=True)
scaler_path = output_dir / "feature_scaler.json"
scaler_dict = {
    "feature_names": all_feature_names,
    "means":         train_means.to_dict(),
    "stds":          train_stds.to_dict(),
    "dataset":       dataset_name,
    "horizon":       horizon,
    "rebalance_every_n_days": rebalance_every_n_days,
}
with open(scaler_path, "w") as f:
    _json.dump(scaler_dict, f, indent=2)
print(f"Scaler saved to: {scaler_path}")

# ── Standardize splits ───────────────────────────────────────────────────────
def standardize(feature_frame: pd.DataFrame) -> np.ndarray:
    standardized = (feature_frame[all_feature_names] - train_means) / train_stds
    return standardized.fillna(0.0).to_numpy(dtype=float)

X_train = standardize(train)
X_val   = standardize(val)
X_test  = standardize(test_rebalance)

y_train_return = train[return_target_col].to_numpy(dtype=float)
y_val_return   = val[return_target_col].to_numpy(dtype=float)
y_test_return  = test_rebalance[return_target_col].to_numpy(dtype=float)

y_train_alpha  = train[alpha_target_col].to_numpy(dtype=float)
y_val_alpha    = val[alpha_target_col].to_numpy(dtype=float)

y_train_vol    = train[vol_target_col].to_numpy(dtype=float)
y_val_vol      = val[vol_target_col].to_numpy(dtype=float)

print("X_train shape:", X_train.shape)
print("X_test shape (rebalanced):", X_test.shape)
print("Feature count:", X_train.shape[1])
print("NaN values after standardization:", np.isnan(X_train).sum())
```

## 6. Train CatBoost Models - With Seed Ensemble

We train ENSEMBLE_SEEDS independent CatBoost models per target:

1. Joint return + alpha (MultiRMSE) - 3 seeds, predictions averaged.
2. Vol rank - 3 seeds, predictions averaged.

Per-prediction uncertainty = std across seeds for each row. This replaces the
previous placeholder (sqrt of overall val MSE which was constant for every row).


```python
from catboost import CatBoostRegressor, Pool

def make_catboost_params(seed):
    return dict(
        iterations=600,
        learning_rate=0.03,
        depth=6,
        l2_leaf_reg=8.0,
        min_data_in_leaf=50,
        bagging_temperature=1.0,
        rsm=0.8,
        random_seed=seed,
        verbose=False,
    )

SEEDS = list(range(42, 42 + ENSEMBLE_SEEDS))

# Joint return + alpha ensemble
y_train_joint = np.column_stack([y_train_return, y_train_alpha])
y_val_joint   = np.column_stack([y_val_return,   y_val_alpha])

joint_models = []
for seed in SEEDS:
    print(f"Training joint model seed={seed}...")
    m = CatBoostRegressor(**make_catboost_params(seed), loss_function="MultiRMSE")
    m.fit(X_train, y_train_joint, eval_set=Pool(X_val, y_val_joint), early_stopping_rounds=40)
    joint_models.append(m)

joint_val_preds = np.stack([m.predict(X_val) for m in joint_models], axis=0)
joint_val_mean  = joint_val_preds.mean(axis=0)
return_val_mse  = float(((joint_val_mean[:, 0] - y_val_return) ** 2).mean())
alpha_val_mse   = float(((joint_val_mean[:, 1] - y_val_alpha)  ** 2).mean())
print(f"Joint ensemble - return MSE: {return_val_mse:.8f} | alpha MSE: {alpha_val_mse:.8f}")
joint_model = joint_models[0]

# Vol rank ensemble
y_train_vol_rank = pd.Series(y_train_vol).rank(pct=True).to_numpy()
y_val_vol_rank   = pd.Series(y_val_vol).rank(pct=True).to_numpy()

vol_models = []
for seed in SEEDS:
    print(f"Training vol model seed={seed}...")
    m = CatBoostRegressor(**make_catboost_params(seed))
    m.fit(X_train, y_train_vol_rank, eval_set=[(X_val, y_val_vol_rank)], early_stopping_rounds=40)
    vol_models.append(m)

vol_val_preds = np.stack([m.predict(X_val) for m in vol_models], axis=0)
vol_val_mean  = vol_val_preds.mean(axis=0)
vol_val_mse   = float(((vol_val_mean - y_val_vol_rank) ** 2).mean())
print(f"Vol ensemble - val MSE: {vol_val_mse:.8f}")
vol_model = vol_models[0]

fi = pd.DataFrame({
    "feature":    all_feature_names,
    "importance": joint_models[0].get_feature_importance()
}).sort_values("importance", ascending=False)
print("\nTop 10 features by importance:")
display(fi.head(10))
print(f"\nEnsemble trained: {len(joint_models)} joint + {len(vol_models)} vol models.")
```

## 9. Turn Predictions Into A Portfolio Object

The toolkit separates forecasting from portfolio construction.

Here we use the built-in `weights_from_predictions_risk_adjusted(...)` helper.

What it does:

- uses `expected_return / expected_volatility` as the score
- keeps the allocation long-only
- normalizes the scores so each row sums to `1.0`
- returns a `PortfolioWeights` object

This is a good default for demonstrations because it uses more of the prediction contract than a plain top-k rule.

```python
# ════════════════════════════════════════════════════════════════════════════
# Portfolio construction + safety rails installation (combined into one cell
# so safety rails are ALWAYS in scope before any weight operations,
# regardless of cell execution order)
# ════════════════════════════════════════════════════════════════════════════
import portfolio_toolkit.validation as _val
import portfolio_toolkit.backtest   as _bt

def _apply_cap_with_redistribution(row, cap):
    """Cap weights at `cap`, redistribute excess proportionally to under-cap names."""
    w = row.astype(float).copy()
    if w.sum() <= 0:
        return w
    w = w / w.sum()
    n_active = (w > 0).sum()
    effective_cap = max(cap, 1.0 / max(n_active, 1))
    for _ in range(100):
        over = w > effective_cap
        if not over.any():
            break
        excess = float((w.loc[over] - effective_cap).sum())
        w.loc[over] = effective_cap
        under = (~over) & (w > 0)
        under_total = float(w.loc[under].sum())
        if under_total <= 0 or excess <= 1e-12:
            break
        w.loc[under] = w.loc[under] + excess * (w.loc[under] / under_total)
    return w / w.sum() if w.sum() > 0 else w

def apply_safety_rails(weights_df, max_cap, min_active):
    """Apply concentration safety rails to a weights DataFrame.
    1. Replace inf/-inf with 0
    2. Zero rows -> equal weight
    3. Renormalize each row to sum to 1
    4. Cap each weight at max_cap with redistribution
    5. If too few active names, fall back to equal weight
    6. Final renormalize
    """
    df = weights_df.copy()
    n_cols = df.shape[1]
    df = df.replace([float("inf"), float("-inf")], 0.0)

    row_sums = df.sum(axis=1)
    zero_rows = row_sums.abs() < 1e-8
    if zero_rows.any():
        df.loc[zero_rows] = 1.0 / n_cols

    df = df.div(df.sum(axis=1), axis=0).fillna(1.0 / n_cols)
    capped = df.apply(lambda r: _apply_cap_with_redistribution(r, max_cap), axis=1)

    active_counts = (capped > 0).sum(axis=1)
    too_few = active_counts < min_active
    if too_few.any():
        capped.loc[too_few] = 1.0 / n_cols

    capped = capped.div(capped.sum(axis=1), axis=0)
    return capped

# Install patched validator (in case the toolkit calls it internally somewhere)
_original_validate = _val.validate_weights_frame
def _patched_validate(df, dataset_name=None, repo_root=None):
    return apply_safety_rails(df, MAX_WEIGHT_CAP, MIN_ACTIVE_NAMES)
_val.validate_weights_frame = _patched_validate
_bt.validate_weights_frame  = _patched_validate

# Build portfolio with toolkit risk-adjusted weight builder
portfolio = weights_from_predictions_risk_adjusted(
    predictions,
    dataset_name=dataset_name,
    strategy_name=model_name,
)

# Apply safety rails DIRECTLY to portfolio.weights so the backtest sees the cap
portfolio.weights = apply_safety_rails(portfolio.weights, MAX_WEIGHT_CAP, MIN_ACTIVE_NAMES)
validated_weights = portfolio.weights.copy()

# Immediate sanity check
_max_check  = float(validated_weights.max(axis=1).max())
_min_active = int((validated_weights > 0).sum(axis=1).min())
print(f"After cap - max single weight: {_max_check:.4f} (target <= {MAX_WEIGHT_CAP})")
print(f"After cap - min active names: {_min_active} (target >= {MIN_ACTIVE_NAMES})")
assert _max_check <= MAX_WEIGHT_CAP + 1e-4, f"Cap not applied! max={_max_check:.4f} > cap={MAX_WEIGHT_CAP}"
assert _min_active >= MIN_ACTIVE_NAMES, f"Min active not enforced! min={_min_active} < target={MIN_ACTIVE_NAMES}"
print("Safety rails verified BEFORE backtest.")

display(validated_weights.head())
```

The toolkit keeps MLflow intentionally lightweight:

The pattern here is the one you want teammates to reuse:

1. initialize MLflow locally
2. start a run with meaningful tags

```python
# ══════════════════════════════════════════════════════════════════════════════
# SUBMISSION BLOCK — Required functions, artifact saving, feature/scaler refs
# ══════════════════════════════════════════════════════════════════════════════

# ── Required Function 1: build_model_features ────────────────────────────────
def build_model_features(prices: pd.DataFrame) -> pd.DataFrame:
    """
    Rebuild the exact feature table used by the model from raw OHLCV price data.
    Accepts: date, ticker, open, high, low, close, adj_close, volume.
    Returns: date, ticker, + all feature columns in all_feature_names order.
    Never uses forward-looking columns (forward_return_*, forward_alpha_*, etc).
    """
    eps = 1e-6

    # ── Base toolkit features ─────────────────────────────────────────────────
    frame = build_features(prices, feature_names=base_feature_names)

    # ── Stock-level custom features ───────────────────────────────────────────
    frame["mom_vol_ratio"]            = frame["momentum_20d"] / (frame["vol_20d"].abs() + eps)
    frame["trend_spread"]             = frame["price_to_sma_20d"] - frame["price_to_sma_50d"]
    frame["quality_signal"]           = frame["excess_return_20d_vs_spy"] - 0.5 * frame["vol_20d"]
    frame["range_volume_interaction"] = frame["intraday_range"] * frame["volume_zscore_20d"]

    # ── Volatility regime features ────────────────────────────────────────────
    daily_returns = (
        prices.sort_values(["ticker", "date"])
        .assign(daily_ret=lambda df: df.groupby("ticker")["close"].pct_change())
    )
    rolling_sharpe = (
        daily_returns.groupby("ticker")["daily_ret"]
        .transform(lambda s: s.rolling(20).mean() / (s.rolling(20).std(ddof=0) + eps))
    )
    daily_returns["rolling_sharpe_20d"] = rolling_sharpe * (252 ** 0.5)
    frame = frame.merge(
        daily_returns[["date", "ticker", "rolling_sharpe_20d"]], on=["date", "ticker"], how="left"
    )
    vol_of_vol = (
        daily_returns.groupby("ticker")["daily_ret"]
        .transform(lambda s: s.rolling(5).std().rolling(20).std())
    )
    daily_returns["vol_of_vol"] = vol_of_vol
    frame = frame.merge(
        daily_returns[["date", "ticker", "vol_of_vol"]], on=["date", "ticker"], how="left"
    )
    drawdown_20d = (
        prices.sort_values(["ticker", "date"])
        .groupby("ticker")["close"]
        .transform(lambda s: s / s.rolling(20).max() - 1)
    )
    drawdown_lookup = (
        prices.sort_values(["ticker", "date"])[["date", "ticker"]]
        .assign(drawdown_20d=drawdown_20d.values)
    )
    frame = frame.merge(drawdown_lookup, on=["date", "ticker"], how="left")

    # ── Macro / regime features ───────────────────────────────────────────────
    # SPY used only as feature context — never as a tradable ticker.
    _start = prices["date"].min().strftime("%Y-%m-%d")
    _end   = prices["date"].max().strftime("%Y-%m-%d")

    def _fetch_close(ticker, start, end):
        df = yf.download(ticker, start=start, end=end, progress=False, auto_adjust=True)
        close = df["Close"]
        if isinstance(close, pd.DataFrame):
            close = close.iloc[:, 0]
        return close

    vix_raw = _fetch_close("^VIX", _start, _end).rename("vix_close")
    tlt_raw = _fetch_close("TLT",  _start, _end).rename("tlt_close")
    macro = (
        pd.concat([vix_raw, tlt_raw], axis=1).reset_index()
        .rename(columns={"Date": "date", "Datetime": "date", "index": "date"})
        .assign(date=lambda df: pd.to_datetime(df["date"]).dt.normalize())
    )
    macro["vix_change_5d"]    = macro["vix_close"].pct_change(5)
    macro["tlt_momentum_20d"] = macro["tlt_close"].pct_change(20)
    macro["rate_regime"]      = macro["tlt_momentum_20d"].apply(lambda x: 1.0 if x >= 0 else -1.0)
    macro = macro[["date", "vix_close", "vix_change_5d", "tlt_momentum_20d", "rate_regime"]].dropna()
    frame["date"] = pd.to_datetime(frame["date"]).dt.normalize()
    frame = frame.merge(macro, on="date", how="left")

    # ── Regime categorical features ───────────────────────────────────────────
    frame["vix_regime"] = pd.cut(
        frame["vix_close"], bins=[0, 15, 20, 30, 100], labels=[0, 1, 2, 3]
    ).astype(float)
    frame["momentum_regime"] = pd.cut(
        frame["momentum_20d"], bins=[-np.inf, -0.05, 0, 0.05, np.inf], labels=[0, 1, 2, 3]
    ).astype(float)
    daily_ret_std5  = daily_returns.groupby("ticker")["daily_ret"].transform(lambda s: s.rolling(5).std())
    daily_ret_std20 = daily_returns.groupby("ticker")["daily_ret"].transform(lambda s: s.rolling(20).std())
    vol_expansion = daily_ret_std5 / (daily_ret_std20 + eps)
    vol_exp_lookup = daily_returns[["date", "ticker"]].assign(vol_expansion_ratio=vol_expansion.values)
    frame = frame.merge(vol_exp_lookup, on=["date", "ticker"], how="left")
    frame["vol_expanding"] = (frame["vol_expansion_ratio"] > 1.2).astype(float)

    # ── Cross-sectional rank features ─────────────────────────────────────────
    for src_col, rank_col in [
        ("return_20d",         "xs_return_rank"),
        ("momentum_20d",       "xs_momentum_rank"),
        ("rolling_sharpe_20d", "xs_sharpe_rank"),
        ("quality_signal",     "xs_quality_rank"),
    ]:
        if src_col in frame.columns:
            frame[rank_col] = frame.groupby("date")[src_col].rank(pct=True)

    return validate_feature_frame(frame)


# ── Required Function 2: predict_from_prices ─────────────────────────────────
# Exact spec signature: predict_from_prices(model, prices, dates=None, tickers=None)
# model = dict with keys "joint_model" and "vol_model", or bare joint_model.
# Scaler is reloaded from the saved feature_scaler.json artifact.
_saved_scaler_path = scaler_path  # captured at definition time

def predict_from_prices(
    model,
    prices: pd.DataFrame,
    dates=None,
    tickers=None,
) -> pd.DataFrame:
    """
    Rebuild features, apply training scaler, run inference, return predictions.
    Fully self-contained — reloads scaler from disk, no live memory needed.
    """
    import json as _j
    with open(_saved_scaler_path) as f:
        _scaler = _j.load(f)
    _means        = pd.Series(_scaler["means"])
    _stds         = pd.Series(_scaler["stds"])
    _feature_names = _scaler["feature_names"]
    _horizon       = _scaler["horizon"]

    # Accept dict {"joint_model": ..., "vol_model": ...} or bare joint_model
    if isinstance(model, dict):
        _models = model
    else:
        from catboost import CatBoostRegressor as _CB
        _vm = _CB()
        _vm.load_model(str(_saved_scaler_path).replace("feature_scaler.json", "vol_model.cbm"))
        _models = {"joint_model": model, "vol_model": _vm}

    features = build_model_features(prices)

    if dates is not None:
        dates = pd.DatetimeIndex(pd.to_datetime(dates))
        features = features.loc[features["date"].isin(dates)].copy()

    if tickers is not None:
        tickers = [str(t).upper() for t in tickers]
        features = features.loc[features["ticker"].isin(tickers)].copy()

    # Drop rows with NaN in any non-rank feature (rank cols filled via groupby)
    non_rank_features = [f for f in _feature_names if not f.startswith("xs_")]
    scoring_frame = features.dropna(subset=non_rank_features).reset_index(drop=True)

    # Apply training scaler
    X = ((scoring_frame[_feature_names] - _means) / _stds).fillna(0.0).to_numpy(dtype=float)

    # Inference
    joint_pred  = _models["joint_model"].predict(X)
    vol_pred    = _models["vol_model"].predict(X).clip(0.01, 0.99)

    predictions = scoring_frame[["date", "ticker"]].copy()
    predictions["horizon"]             = _horizon
    predictions["expected_return"]     = joint_pred[:, 0]
    predictions["expected_alpha"]      = joint_pred[:, 1]
    predictions["expected_volatility"] = vol_pred

    return validate_prediction_frame(predictions, horizon=_horizon)


# ── Save model artifacts (.cbm format for CatBoost) ──────────────────────────
output_dir.mkdir(parents=True, exist_ok=True)
joint_model_path = output_dir / "joint_model.cbm"
vol_model_path   = output_dir / "vol_model.cbm"
joint_model.save_model(str(joint_model_path), format="cbm")
vol_model.save_model(str(vol_model_path),   format="cbm")

print("✓ build_model_features defined")
print("✓ predict_from_prices defined")
print(f"✓ joint_model saved: {joint_model_path}")
print(f"✓ vol_model saved:   {vol_model_path}")
print(f"✓ scaler saved:      {scaler_path}")
print(f"✓ feature_names: {len(all_feature_names)} features | horizon: {horizon}d | rebalance: every_{rebalance_every_n_days}_trading_days")
```

```python
def _felix_run13_sort_runs_latest_first(runs: pd.DataFrame) -> pd.DataFrame:
    if runs.empty:
        return runs
    sorted_runs = runs.copy()
    if "start_time" in sorted_runs.columns:
        sorted_runs = sorted_runs.sort_values("start_time", ascending=False)
    return sorted_runs.reset_index(drop=True)


def _felix_run13_find_latest_mlflow_run_id() -> str:
    forced_run_id = os.environ.get("FELIX_RUN13_MLFLOW_RUN_ID", "").strip()
    if forced_run_id:
        return forced_run_id

    init_mlflow(repo_root=repo_root)
    experiments = mlflow.search_experiments()
    experiment_ids = [experiment.experiment_id for experiment in experiments]
    if not experiment_ids:
        raise RuntimeError("MLflow has no experiments visible to search for Felix Run13.")

    runs = pd.DataFrame()
    try:
        runs = mlflow.search_runs(
            experiment_ids=experiment_ids,
            filter_string=f"params.submission_model_name = '{FELIX_RUN13_MODEL_NAME}'",
            max_results=1000,
        )
    except Exception as exc:
        print(f"Filtered MLflow search failed, falling back to unfiltered search: {exc}")

    if runs.empty:
        runs = mlflow.search_runs(experiment_ids=experiment_ids, max_results=3000)
    if runs.empty:
        raise RuntimeError("MLflow search returned no runs.")

    candidate_mask = pd.Series(False, index=runs.index)
    for column in ["params.submission_model_name", "params.model_name", "tags.mlflow.runName"]:
        if column in runs.columns:
            candidate_mask = candidate_mask | runs[column].astype(str).str.contains(FELIX_RUN13_MODEL_NAME, case=False, na=False)
    candidates = runs.loc[candidate_mask].copy()
    if candidates.empty:
        raise RuntimeError(
            f"Could not find an MLflow run for {FELIX_RUN13_MODEL_NAME!r}. "
            "Set FELIX_RUN13_MLFLOW_RUN_ID to the exact run id."
        )

    run_name = candidates.get("tags.mlflow.runName", pd.Series("", index=candidates.index)).astype(str)
    run13_mask = run_name.str.contains(FELIX_RUN13_MLFLOW_RUN_NAME_HINT, case=False, na=False) | run_name.str.contains("run13", case=False, na=False)
    if run13_mask.any():
        candidates = candidates.loc[run13_mask].copy()
    else:
        print("Warning: no run name matched Run13; using the latest catboost_sp500 model submission visible in MLflow.")

    candidates = _felix_run13_sort_runs_latest_first(candidates)
    return str(candidates.loc[0, "run_id"])


def _felix_run13_list_artifacts_recursive(client: MlflowClient, run_id: str, path: str = "") -> list[str]:
    artifact_paths: list[str] = []
    for artifact in client.list_artifacts(run_id, path):
        if artifact.is_dir:
            artifact_paths.extend(_felix_run13_list_artifacts_recursive(client, run_id, artifact.path))
        else:
            artifact_paths.append(artifact.path)
    return artifact_paths


def _felix_run13_download_artifact(run_id: str, artifact_path: str) -> Path:
    local_path = Path(mlflow.artifacts.download_artifacts(run_id=run_id, artifact_path=artifact_path))
    if not local_path.exists():
        raise FileNotFoundError(f"Downloaded artifact path does not exist: {local_path}")
    return local_path


def _felix_run13_load_manifest(run_id: str, artifact_paths: list[str]) -> tuple[dict, str | None]:
    manifest_paths = [path for path in artifact_paths if path.endswith("manifest.json") and "model_submission" in path]
    if not manifest_paths:
        return {}, None
    manifest_path = sorted(manifest_paths, key=len)[0]
    local_manifest_path = _felix_run13_download_artifact(run_id, manifest_path)
    return json.loads(local_manifest_path.read_text(encoding="utf-8")), manifest_path


def _felix_run13_artifact_path_from_manifest(manifest: dict, manifest_path: str, logical_name: str) -> str:
    artifact_map = manifest.get("artifact_map", {})
    if logical_name not in artifact_map:
        raise KeyError(f"Manifest artifact_map is missing {logical_name!r}: {artifact_map}")
    manifest_parent = Path(manifest_path).parent.as_posix()
    return f"{manifest_parent}/{artifact_map[logical_name]}"


def _felix_run13_scaler_from_manifest(run_id: str, manifest: dict, scaler_artifact_path: str | None) -> tuple[dict, Path, str]:
    if scaler_artifact_path:
        scaler_path = _felix_run13_download_artifact(run_id, scaler_artifact_path)
        return json.loads(scaler_path.read_text(encoding="utf-8")), scaler_path, scaler_artifact_path

    preprocessing = manifest.get("preprocessing", {})
    feature_names = list(manifest.get("feature_names", []))
    if not feature_names or "means" not in preprocessing or "stds" not in preprocessing:
        raise RuntimeError("Could not find feature_scaler.json and the manifest does not contain reconstructable mean/std preprocessing.")

    scaler = {
        "feature_names": feature_names,
        "means": preprocessing["means"],
        "stds": preprocessing["stds"],
        "dataset": "mlflow_manifest_reconstructed",
        "horizon": int(manifest.get("horizon", FELIX_RUN13_HORIZON)),
        "rebalance_every_n_days": FELIX_RUN13_REBALANCE_EVERY_N_DAYS,
    }
    reconstructed_path = FELIX_RUN13_REGIME_BACKTEST_OUTPUT_DIR / "reconstructed_feature_scaler_from_manifest.json"
    reconstructed_path.write_text(json.dumps(scaler, indent=2, sort_keys=True) + "\n", encoding="utf-8")
    return scaler, reconstructed_path, "reconstructed_from_manifest_preprocessing"


def _felix_run13_load_model_bundle_from_mlflow() -> dict[str, object]:
    init_mlflow(repo_root=repo_root)
    run_id = _felix_run13_find_latest_mlflow_run_id()
    client = MlflowClient()
    artifact_paths = _felix_run13_list_artifacts_recursive(client, run_id)
    manifest, manifest_path = _felix_run13_load_manifest(run_id, artifact_paths)

    forced_joint = os.environ.get("FELIX_RUN13_MLFLOW_ARTIFACT_PATH_JOINT", "").strip()
    forced_vol = os.environ.get("FELIX_RUN13_MLFLOW_ARTIFACT_PATH_VOL", "").strip()
    forced_scaler = os.environ.get("FELIX_RUN13_MLFLOW_ARTIFACT_PATH_SCALER", "").strip()

    if forced_joint and forced_vol:
        joint_artifact_path = forced_joint
        vol_artifact_path = forced_vol
    elif manifest and manifest_path:
        joint_artifact_path = _felix_run13_artifact_path_from_manifest(manifest, manifest_path, "joint_model")
        vol_artifact_path = _felix_run13_artifact_path_from_manifest(manifest, manifest_path, "vol_model")
    else:
        cbm_paths = sorted([path for path in artifact_paths if path.endswith(".cbm")])
        joint_candidates = [path for path in cbm_paths if "joint" in Path(path).name.lower()]
        vol_candidates = [path for path in cbm_paths if "vol" in Path(path).name.lower()]
        if not joint_candidates or not vol_candidates:
            raise RuntimeError(f"Could not infer Felix Run13 CatBoost artifacts from MLflow paths: {artifact_paths}")
        joint_artifact_path = joint_candidates[0]
        vol_artifact_path = vol_candidates[0]

    if forced_scaler:
        scaler_artifact_path = forced_scaler
    else:
        scaler_candidates = [path for path in artifact_paths if path.endswith("feature_scaler.json")]
        scaler_artifact_path = sorted(scaler_candidates, key=len)[0] if scaler_candidates else None

    joint_path = _felix_run13_download_artifact(run_id, joint_artifact_path)
    vol_path = _felix_run13_download_artifact(run_id, vol_artifact_path)
    scaler, scaler_path, scaler_source = _felix_run13_scaler_from_manifest(run_id, manifest, scaler_artifact_path)

    joint_model = CatBoostRegressor()
    joint_model.load_model(str(joint_path))
    vol_model = CatBoostRegressor()
    vol_model.load_model(str(vol_path))

    feature_names = list(scaler.get("feature_names", manifest.get("feature_names", FELIX_RUN13_ALL_FEATURE_NAMES)))
    if feature_names != FELIX_RUN13_ALL_FEATURE_NAMES:
        print("Warning: MLflow scaler feature order differs from the notebook's Run13 feature list. Using MLflow scaler order for inference.")

    model_info = {
        "run_id": run_id,
        "joint_artifact_path": joint_artifact_path,
        "vol_artifact_path": vol_artifact_path,
        "scaler_artifact_path": scaler_source,
        "local_joint_path": str(joint_path),
        "local_vol_path": str(vol_path),
        "local_scaler_path": str(scaler_path),
    }
    bundle = {
        "joint_model": joint_model,
        "vol_model": vol_model,
        "scaler": scaler,
        "manifest": manifest,
        "model_info": model_info,
        "feature_names": feature_names,
    }
    print("Loaded Felix Run13 model bundle from MLflow run:", run_id)
    print("Joint artifact:", joint_artifact_path)
    print("Vol artifact:", vol_artifact_path)
    print("Scaler artifact:", scaler_source)
    return bundle
```

```python
def _felix_run13_normalize_yfinance_frame(frame: pd.DataFrame, ticker: str) -> pd.DataFrame:
    normalized = frame.copy()
    if isinstance(normalized.columns, pd.MultiIndex):
        normalized.columns = normalized.columns.get_level_values(0)
    normalized.columns.name = None
    if "Adj Close" not in normalized.columns and "Close" in normalized.columns:
        normalized["Adj Close"] = normalized["Close"]
    normalized = normalized.reset_index().rename(
        columns={
            "Date": "date",
            "Datetime": "date",
            "Open": "open",
            "High": "high",
            "Low": "low",
            "Close": "close",
            "Adj Close": "adj_close",
            "Volume": "volume",
        }
    )
    normalized["ticker"] = ticker.upper()
    normalized = normalized.loc[:, ["date", "ticker", "open", "high", "low", "close", "adj_close", "volume"]]
    normalized["date"] = pd.to_datetime(normalized["date"], utc=True).dt.tz_localize(None)
    for column in ["open", "high", "low", "close", "adj_close", "volume"]:
        normalized[column] = pd.to_numeric(normalized[column], errors="coerce")
    return normalized.dropna(subset=["date", "adj_close"]).sort_values(["ticker", "date"]).reset_index(drop=True)


def _felix_run13_download_context_ticker(ticker: str, spec_obj) -> pd.DataFrame:
    start = pd.Timestamp(spec_obj.start_date).date().isoformat()
    end = (pd.Timestamp(spec_obj.end_date).date() + timedelta(days=1)).isoformat()
    downloaded = yf.download(ticker, start=start, end=end, auto_adjust=False, progress=False, threads=False)
    if downloaded.empty:
        raise ValueError(f"No yfinance rows downloaded for required context ticker {ticker} in {spec_obj.identifier}.")
    return _felix_run13_normalize_yfinance_frame(downloaded, ticker)


def _felix_run13_prices_with_spy_context(prices_frame: pd.DataFrame, spec_obj) -> pd.DataFrame:
    prices_for_features = prices_frame.copy()
    prices_for_features["date"] = pd.to_datetime(prices_for_features["date"], utc=True).dt.tz_localize(None)
    prices_for_features["ticker"] = prices_for_features["ticker"].astype(str).str.upper()
    if "SPY" not in set(prices_for_features["ticker"]):
        spy_context = _felix_run13_download_context_ticker("SPY", spec_obj)
        prices_for_features = pd.concat([prices_for_features, spy_context], ignore_index=True)
    return prices_for_features.sort_values(["ticker", "date"]).reset_index(drop=True)


def _felix_run13_fetch_close(ticker: str, start: pd.Timestamp, end: pd.Timestamp) -> pd.Series:
    try:
        downloaded = yf.download(
            ticker,
            start=pd.Timestamp(start).date().isoformat(),
            end=(pd.Timestamp(end).date() + timedelta(days=1)).isoformat(),
            progress=False,
            auto_adjust=True,
            threads=False,
        )
    except Exception as exc:
        print(f"Warning: failed to download {ticker} context data: {exc}")
        return pd.Series(dtype=float, name=ticker)
    if downloaded.empty or "Close" not in downloaded.columns:
        return pd.Series(dtype=float, name=ticker)
    close = downloaded["Close"]
    if isinstance(close, pd.DataFrame):
        close = close.iloc[:, 0]
    close = pd.to_numeric(close, errors="coerce")
    close.index = pd.to_datetime(close.index, utc=True).tz_localize(None).normalize()
    close.name = ticker
    return close.dropna()


def _felix_run13_build_model_features(prices_frame: pd.DataFrame) -> pd.DataFrame:
    eps = 1e-6
    prices_local = prices_frame.copy()
    prices_local["date"] = pd.to_datetime(prices_local["date"], utc=True).dt.tz_localize(None)
    prices_local["ticker"] = prices_local["ticker"].astype(str).str.upper()
    prices_local = prices_local.sort_values(["ticker", "date"]).reset_index(drop=True)

    frame = build_features(prices_local, feature_names=FELIX_RUN13_BASE_FEATURE_NAMES)
    frame["mom_vol_ratio"] = frame["momentum_20d"] / (frame["vol_20d"].abs() + eps)
    frame["trend_spread"] = frame["price_to_sma_20d"] - frame["price_to_sma_50d"]
    frame["quality_signal"] = frame["excess_return_20d_vs_spy"] - 0.5 * frame["vol_20d"]
    frame["range_volume_interaction"] = frame["intraday_range"] * frame["volume_zscore_20d"]

    daily_returns = prices_local.sort_values(["ticker", "date"]).assign(
        daily_ret=lambda df: df.groupby("ticker")["close"].pct_change()
    )
    rolling_sharpe = daily_returns.groupby("ticker")["daily_ret"].transform(
        lambda s: s.rolling(20).mean() / (s.rolling(20).std(ddof=0) + eps)
    )
    daily_returns["rolling_sharpe_20d"] = rolling_sharpe * (252 ** 0.5)
    frame = frame.merge(daily_returns[["date", "ticker", "rolling_sharpe_20d"]], on=["date", "ticker"], how="left")

    daily_returns["vol_of_vol"] = daily_returns.groupby("ticker")["daily_ret"].transform(
        lambda s: s.rolling(5).std().rolling(20).std()
    )
    frame = frame.merge(daily_returns[["date", "ticker", "vol_of_vol"]], on=["date", "ticker"], how="left")

    drawdown_20d = prices_local.groupby("ticker")["close"].transform(lambda s: s / s.rolling(20).max() - 1)
    drawdown_lookup = prices_local[["date", "ticker"]].assign(drawdown_20d=drawdown_20d.values)
    frame = frame.merge(drawdown_lookup, on=["date", "ticker"], how="left")

    start = pd.to_datetime(prices_local["date"]).min()
    end = pd.to_datetime(prices_local["date"]).max()
    vix_raw = _felix_run13_fetch_close("^VIX", start, end).rename("vix_close")
    tlt_raw = _felix_run13_fetch_close("TLT", start, end).rename("tlt_close")
    macro = pd.concat([vix_raw, tlt_raw], axis=1).reset_index().rename(
        columns={"index": "date", "Date": "date", "Datetime": "date"}
    )
    if macro.empty or "date" not in macro.columns:
        macro = pd.DataFrame({"date": pd.to_datetime(frame["date"]).drop_duplicates().sort_values()})
        macro["vix_close"] = np.nan
        macro["tlt_close"] = np.nan
    macro["date"] = pd.to_datetime(macro["date"], utc=True).dt.tz_localize(None).dt.normalize()
    if "vix_close" not in macro.columns:
        macro["vix_close"] = np.nan
    if "tlt_close" not in macro.columns:
        macro["tlt_close"] = np.nan
    macro["vix_change_5d"] = macro["vix_close"].pct_change(5)
    macro["tlt_momentum_20d"] = macro["tlt_close"].pct_change(20)
    macro["rate_regime"] = np.where(
        macro["tlt_momentum_20d"].notna(),
        np.where(macro["tlt_momentum_20d"] >= 0, 1.0, -1.0),
        np.nan,
    )
    macro = macro[["date", "vix_close", "vix_change_5d", "tlt_momentum_20d", "rate_regime"]]
    frame["date"] = pd.to_datetime(frame["date"], utc=True).dt.tz_localize(None).dt.normalize()
    frame = frame.merge(macro, on="date", how="left")

    frame["vix_regime"] = pd.cut(frame["vix_close"], bins=[0, 15, 20, 30, 100], labels=[0, 1, 2, 3]).astype(float)
    frame["momentum_regime"] = pd.cut(
        frame["momentum_20d"], bins=[-np.inf, -0.05, 0, 0.05, np.inf], labels=[0, 1, 2, 3]
    ).astype(float)
    daily_ret_std5 = daily_returns.groupby("ticker")["daily_ret"].transform(lambda s: s.rolling(5).std())
    daily_ret_std20 = daily_returns.groupby("ticker")["daily_ret"].transform(lambda s: s.rolling(20).std())
    vol_expansion = daily_ret_std5 / (daily_ret_std20 + eps)
    frame = frame.merge(
        daily_returns[["date", "ticker"]].assign(vol_expansion_ratio=vol_expansion.values),
        on=["date", "ticker"],
        how="left",
    )
    frame["vol_expanding"] = (frame["vol_expansion_ratio"] > 1.2).astype(float)

    for src_col, rank_col in [
        ("return_20d", "xs_return_rank"),
        ("momentum_20d", "xs_momentum_rank"),
        ("rolling_sharpe_20d", "xs_sharpe_rank"),
        ("quality_signal", "xs_quality_rank"),
    ]:
        frame[rank_col] = frame.groupby("date")[src_col].rank(pct=True) if src_col in frame.columns else np.nan

    for column in FELIX_RUN13_ALL_FEATURE_NAMES:
        if column not in frame.columns:
            frame[column] = np.nan
    return validate_feature_frame(frame[["date", "ticker"] + FELIX_RUN13_ALL_FEATURE_NAMES])


def _felix_run13_predict_from_prices(model_bundle: dict[str, object], prices: pd.DataFrame, dates=None, tickers=None) -> pd.DataFrame:
    scaler = model_bundle["scaler"]
    feature_names = list(model_bundle.get("feature_names") or scaler.get("feature_names") or FELIX_RUN13_ALL_FEATURE_NAMES)
    horizon = int(scaler.get("horizon", FELIX_RUN13_HORIZON))
    means = pd.to_numeric(pd.Series(scaler.get("means", {})).reindex(feature_names), errors="coerce").fillna(0.0)
    stds = pd.to_numeric(pd.Series(scaler.get("stds", {})).reindex(feature_names), errors="coerce").replace(0.0, 1.0).fillna(1.0)

    features = _felix_run13_build_model_features(prices)
    if dates is not None:
        requested_dates = pd.DatetimeIndex(pd.to_datetime(dates, utc=True).tz_localize(None)).normalize()
        features = features.loc[features["date"].isin(requested_dates)].copy()
    if tickers is not None:
        requested_tickers = [str(ticker).upper() for ticker in tickers]
        features = features.loc[features["ticker"].isin(requested_tickers)].copy()
    if features.empty:
        raise ValueError("No Felix Run13 feature rows remain after date/ticker filtering.")

    scoring_frame = features.reset_index(drop=True)
    X = ((scoring_frame.reindex(columns=feature_names) - means) / stds).replace([np.inf, -np.inf], np.nan).fillna(0.0)
    X_np = X.to_numpy(dtype=float)

    joint_pred = np.asarray(model_bundle["joint_model"].predict(X_np), dtype=float)
    if joint_pred.ndim == 1:
        expected_return = joint_pred
        expected_alpha = np.full(len(scoring_frame), np.nan)
    else:
        expected_return = joint_pred[:, 0]
        expected_alpha = joint_pred[:, 1] if joint_pred.shape[1] > 1 else np.full(len(scoring_frame), np.nan)

    vol_pred = np.asarray(model_bundle["vol_model"].predict(X_np), dtype=float).reshape(-1)
    vol_pred = np.clip(vol_pred, 0.01, 0.99)

    predictions = scoring_frame[["date", "ticker"]].copy()
    predictions["horizon"] = horizon
    predictions["expected_return"] = expected_return
    predictions["expected_alpha"] = expected_alpha
    predictions["expected_volatility"] = vol_pred
    predictions["uncertainty"] = 1e-6
    return validate_prediction_frame(predictions, horizon=horizon)
```

```python
def _felix_run13_apply_cap_with_redistribution(row: pd.Series, cap: float) -> pd.Series:
    weights = row.astype(float).replace([np.inf, -np.inf], 0.0).fillna(0.0)
    if weights.sum() <= 0:
        return weights
    weights = weights / weights.sum()
    effective_cap = max(float(cap), 1.0 / max(int((weights > 0).sum()), 1))
    for _ in range(100):
        over = weights > effective_cap
        if not over.any():
            break
        excess = float((weights.loc[over] - effective_cap).sum())
        weights.loc[over] = effective_cap
        under = (~over) & (weights > 0)
        under_total = float(weights.loc[under].sum())
        if under_total <= 0 or excess <= 1e-12:
            break
        weights.loc[under] = weights.loc[under] + excess * (weights.loc[under] / under_total)
    return weights / weights.sum() if weights.sum() > 0 else weights


def _felix_run13_apply_safety_rails(weights_df: pd.DataFrame, max_cap: float, min_active: int) -> tuple[pd.DataFrame, dict[str, float]]:
    if weights_df.empty:
        raise ValueError("Cannot apply safety rails to an empty weights frame.")
    df = weights_df.copy().replace([np.inf, -np.inf], 0.0).fillna(0.0)
    n_cols = int(df.shape[1])
    if n_cols <= 0:
        raise ValueError("Weights frame has no ticker columns.")

    effective_min_active = max(1, min(int(min_active), n_cols))
    effective_max_cap = max(float(max_cap), 1.0 / float(effective_min_active))

    zero_rows = df.sum(axis=1).abs() < 1e-12
    if zero_rows.any():
        df.loc[zero_rows] = 1.0 / n_cols
    df = df.div(df.sum(axis=1).replace(0.0, np.nan), axis=0).fillna(1.0 / n_cols)
    capped = df.apply(lambda row: _felix_run13_apply_cap_with_redistribution(row, effective_max_cap), axis=1)

    active_counts = (capped > 1e-12).sum(axis=1)
    too_few = active_counts < effective_min_active
    if too_few.any():
        capped.loc[too_few] = 1.0 / n_cols
    capped = capped.div(capped.sum(axis=1).replace(0.0, np.nan), axis=0).fillna(1.0 / n_cols)
    capped = validate_weights_frame(capped)

    diagnostics = {
        "original_max_weight_cap": float(max_cap),
        "effective_max_weight_cap": float(effective_max_cap),
        "original_min_active_names": float(min_active),
        "effective_min_active_names": float(effective_min_active),
        "available_ticker_count": float(n_cols),
        "fallback_equal_weight_rows": float(too_few.sum()),
        "avg_active_names": float((capped > 1e-12).sum(axis=1).mean()),
        "min_active_names": float((capped > 1e-12).sum(axis=1).min()),
        "avg_max_weight": float(capped.max(axis=1).mean()),
        "max_single_weight": float(capped.max(axis=1).max()),
    }
    return capped, diagnostics


def _felix_run13_save_json(payload: dict, path: Path) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(payload, indent=2, sort_keys=True, default=str) + "\n", encoding="utf-8")


def _run_felix_run13_regime_backtest(dataset_name: str, model_bundle: dict[str, object]) -> dict[str, object]:
    spec_obj = get_dataset_spec(dataset_name, repo_root=repo_root)
    benchmark = spec_obj.benchmark_ticker.upper()
    tradable = sorted([ticker.upper() for ticker in spec_obj.tickers if ticker.upper() != benchmark])
    if not tradable:
        raise ValueError(f"{dataset_name} has no tradable tickers after excluding the benchmark.")

    loaded_prices = load_prices(spec_obj, repo_root=repo_root)
    loaded_prices = loaded_prices.copy()
    loaded_prices["date"] = pd.to_datetime(loaded_prices["date"], utc=True).dt.tz_localize(None)
    loaded_prices["ticker"] = loaded_prices["ticker"].astype(str).str.upper()
    available_tradable = sorted(set(tradable).intersection(set(loaded_prices["ticker"])))
    if not available_tradable:
        raise ValueError(f"{dataset_name} has no tradable tickers with loaded prices.")

    feature_prices = _felix_run13_prices_with_spy_context(loaded_prices, spec_obj)
    feature_frame = _felix_run13_build_model_features(feature_prices)
    test_start = pd.Timestamp(spec_obj.test_start)
    test_end = pd.Timestamp(spec_obj.test_end)
    test_feature_rows = feature_frame.loc[
        feature_frame["ticker"].isin(available_tradable)
        & (feature_frame["date"] >= test_start)
        & (feature_frame["date"] <= test_end)
    ]
    test_dates = pd.DatetimeIndex(test_feature_rows["date"].drop_duplicates().sort_values())
    if test_dates.empty:
        raise ValueError(f"No test feature dates found for {dataset_name}: {spec_obj.test_start} to {spec_obj.test_end}")
    rebalance_dates = test_dates[::FELIX_RUN13_REBALANCE_EVERY_N_DAYS]

    predictions = _felix_run13_predict_from_prices(
        model_bundle,
        feature_prices,
        dates=rebalance_dates,
        tickers=available_tradable,
    )
    predictions = validate_prediction_frame(predictions, dataset_name=spec_obj, horizon=FELIX_RUN13_HORIZON, repo_root=repo_root)

    portfolio = weights_from_predictions_risk_adjusted(
        predictions,
        dataset_name=spec_obj,
        strategy_name=f"felix_run13_{dataset_name}",
    )
    raw_weights = portfolio.weights.reindex(columns=available_tradable, fill_value=0.0)
    safe_weights, safety_diagnostics = _felix_run13_apply_safety_rails(
        raw_weights,
        FELIX_RUN13_MAX_WEIGHT_CAP,
        FELIX_RUN13_MIN_ACTIVE_NAMES,
    )
    portfolio = PortfolioWeights(
        weights=safe_weights,
        dataset_name=spec_obj.identifier,
        strategy_name=f"felix_run13_{dataset_name}",
        metadata={
            "model_name": FELIX_RUN13_MODEL_NAME,
            "model_version": "run13_remake",
            "portfolio_builder": "weights_from_predictions_risk_adjusted_plus_run13_safety_rails",
            "rebalance_every_n_days": FELIX_RUN13_REBALANCE_EVERY_N_DAYS,
            **safety_diagnostics,
        },
    )

    result = backtest_weights(spec_obj, portfolio, benchmark=spec_obj.benchmark_ticker, repo_root=repo_root)
    dataset_output_dir = FELIX_RUN13_REGIME_BACKTEST_OUTPUT_DIR / dataset_name
    artifact_paths = write_backtest_artifacts(result, dataset_output_dir)
    predictions_path = dataset_output_dir / "predictions.parquet"
    safety_path = dataset_output_dir / "safety_diagnostics.json"
    predictions.to_parquet(predictions_path, index=False)
    _felix_run13_save_json(safety_diagnostics, safety_path)
    artifact_paths["predictions"] = str(predictions_path)
    artifact_paths["safety_diagnostics"] = str(safety_path)

    print(
        f"{dataset_name}: {len(rebalance_dates)} rebalance dates, "
        f"{len(available_tradable)} tradable tickers, total return {result.metrics.get('total_return', np.nan):.2%}"
    )
    return {
        "dataset_name": dataset_name,
        "spec": spec_obj,
        "prices": loaded_prices,
        "feature_row_count": len(feature_frame),
        "predictions": predictions,
        "portfolio": portfolio,
        "result": result,
        "artifact_paths": artifact_paths,
        "rebalance_count": int(len(rebalance_dates)),
        "active_ticker_count": int(len(available_tradable)),
        "safety_diagnostics": safety_diagnostics,
    }
```
